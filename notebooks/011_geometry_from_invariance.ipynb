The program briefly accumulated attractive geometric slogans—most memorably “spheres all the way down.” Derivation quickly made that slogan untenable as a design rule. Different semantic roles imply different invariances and growth laws.


## A representative pruning result

Suppose `x = r u`, with `||u||=1`, and we want direction to encode semantic identity while radius independently encodes structural depth. Ordinary Euclidean geometry does **not** make these channels independent:

`ds² = dr² + r² dΩ²`

and raw dot products contain `r_q r_k cos(theta)`. Radius changes angular metric scale and similarity. If independence is required, a product geometry or explicit side channel is a cleaner construction; if coupling is intended, a warped geometry may be appropriate.

Other branches received similar decision rules: hyperbolic structure only when growth/separation assumptions justify it; Grassmann/subspace semantics only when persistent contextual rank and basis-gauge invariance matter; noncommutative semantic operations cannot all be represented by a commuting operator family.


```python
import numpy as np
theta=np.pi/3
for rq,rk in [(1,1),(2,1),(2,3)]:
    print(rq,rk,'dot-like similarity=',rq*rk*np.cos(theta))
```

## What this does not show

These reductions do not establish one globally correct semantic geometry. They turn aesthetic architecture choices into conditional statements: **if** an invariance, growth law, task kernel, or operator algebra is required, **then** some geometries are compatible and others are not.

## Understanding questions

**Why aren't radius and angle independent in ordinary Euclidean polar coordinates?**  
Angular distance is scaled by radius and raw inner products multiply both radii into angular similarity.

**When is a product geometry appropriate?**  
When two factors should vary independently under the task metric/operations.

**What became the design principle?**  
Derive geometry from invariance, growth law, task kernel, and operator algebra rather than choosing it because the picture is appealing.
