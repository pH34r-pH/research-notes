A semantic-compiler branch began with a provocative question: could a constrained language such as Lojban act like a better intermediate representation for reasoning? The first correction was important: tokenization and semantic normalization are different operations. A semantic IR should be judged by what distinctions it preserves for future computation.


## From static sufficiency to dynamic sufficiency

For a single task, states can be quotiented when they are indistinguishable for that task. Recurrent reasoning requires more: if two states are considered equivalent, the recurrence must send them to states that remain equivalent. Otherwise the quotient is not closed under the dynamics.

> **Sticky note — quotient:** A quotient groups objects into equivalence classes and treats members of the same class as the same object for the declared purpose.

? The resulting research target became a **dynamic/predictive quotient**: preserve exactly the distinctions required for future observations/actions, rather than maximizing raw retained information.


```python
# Toy quotient: states are equivalent if they have the same parity.
states=range(6)
R=lambda x:(x+2)%6  # respects parity
for x in states:
    print(x, 'class', x%2, '->', R(x), 'class', R(x)%2)
```

## What this does not show

A dynamic quotient theorem does not tell us which equivalence relation a language model should learn. Nor does a lossy semantic compiler become universally sufficient: future tasks may depend on distinctions the compiler discarded.

## Understanding questions

**Why is static task equivalence insufficient for recurrence?**  
Equivalent states can evolve into distinguishable future states unless the recurrence respects the equivalence relation.

**Why not preserve all information?**  
The goal is task/prediction-sufficient state, and retaining irrelevant degrees of freedom can increase cost or expose shortcuts.

**What changed conceptually?**  
The representation target became future-behavioral sufficiency rather than a particular serialization language.
