The project accumulated many plausible geometries: spheres, hyperbolic hierarchy, product spaces, subspace-valued concepts, typed operators, sheaf-like transport, and more. Implementing each idea would have turned mathematical uncertainty into an expensive architecture search.

## Four evidence types

✓ **Formal checkpoint** — a mathematical statement proved under explicit assumptions.  
● **Observation** — an empirical measurement under a stated protocol.  
◇ **Assumption** — a premise required by an argument but not established there.  
? **Hypothesis** — a proposed explanation that remains testable.

The private theorem ledger tracks research lifecycle, correspondence, provenance, dependencies, corrections, and experiment consequences. The public [theorem-library](https://github.com/pH34r-pH/theorem-library) exposes reusable Lean mathematics and intrinsic proof validation. This repository is neither source of truth.


```python
claims = {
    'normalization radial derivative is zero': 'formal',
    'unit model had lower NLL at #164 endpoint': 'observation',
    'radial direction is nuisance': 'hypothesis',
}
for claim, kind in claims.items(): print(f'{kind:11s} | {claim}')
```

## What this does not show

Formalization does not make empirical ML deductive. It establishes mathematical implications and impossibility/boundary results; empirical premises still require experiments. A Lean proof also does not automatically promote the surrounding scientific interpretation.

## Understanding questions

**Why keep theorem and observation statuses separate?**  
They are justified by different evidence and fail in different ways.

**What is the practical value of a proof-first ledger?**  
It can eliminate impossible or redundant experimental branches before compute is spent.

**Why isn't research-notes the ledger?**  
It is a curated educational presentation; lifecycle/provenance authority remains with the private research system.
