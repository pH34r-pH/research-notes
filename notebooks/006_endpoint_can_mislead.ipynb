## What this does not show

## Understanding questions

**Why freeze a stopping policy before comparison?**  
Checkpoint selection can otherwise become an outcome-dependent hidden hyperparameter.

**Why does the independent sampling unit matter?**  
Treating dependent rows as independent produces unjustifiably narrow uncertainty.

**What is the research lesson?**  
A spectacular anomaly is a reason to design stronger controls, not a reason to protect the original explanation.
