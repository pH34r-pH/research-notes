# Milestone 003 — Coordinates matter; operations matter more

**Research period:** September 1, 2026  
**Historical anchors:** #161–#162

The next experiment held the uncomposed spectral source state fixed and changed only how downstream computation represented and operated on it.

The crucial control was a parameter-free coordinate rewrite `[rho, cos(theta), sin(theta)]`: if that alone explained the gain, we would have evidence for reparameterization, not a special matched operation.

> **Sticky note — log-polar:** A complex value can be described by magnitude and phase. `rho = log(|z|+eps)` stores magnitude on a logarithmic scale; `theta` stores angle.


## Result

● Polar-native vs spectral Cartesian: about **-0.5825 nat/original byte**.  
● Polar-native vs bit-identical coordinate-only control: about **-0.3923 nat/original byte**.

All three seeds cleared the frozen materiality threshold. Coordinate reparameterization helped, but the matched processing path helped substantially more.

This was the first direct positive evidence in the program that the same representation could behave differently under computation that respected its structure.


```python
import numpy as np
z = np.array([1+1j, -2+0.5j, 0.2-3j])
rho = np.log(np.abs(z) + 1e-8)
phase_features = np.c_[rho, np.cos(np.angle(z)), np.sin(np.angle(z))]
print(phase_features)
```

## What this does not show

One favorable pair does not establish a universal representation–architecture co-design law. It also does not identify which operation inside the polar-native block caused the gain.

## Understanding questions

**Why include a coordinate-only control?**  
To distinguish benefit from an invertible/redundant coordinate rewrite from benefit caused by the operations applied in those coordinates.

**Why hold the source representation fixed?**  
So the contrast targets downstream geometry rather than a newly tuned representation.

**What should happen next?**  
Ablate the matched block until the smallest surviving operation is identified.
